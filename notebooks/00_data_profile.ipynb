```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

RAW_DATA_DIR = PROJECT_ROOT / "data" / "raw"

print("Project root:", PROJECT_ROOT)
print("Raw data folder:", RAW_DATA_DIR)
print("Files found:")

for file in RAW_DATA_DIR.glob("*.csv"):
    print("-", file.name)
```

```python
sales = pd.read_csv(RAW_DATA_DIR / "sales_transactions.csv")
products = pd.read_csv(RAW_DATA_DIR / "products_master.csv")
inventory = pd.read_csv(RAW_DATA_DIR / "inventory_daily.csv")

print("Sales shape:", sales.shape)
print("Products shape:", products.shape)
print("Inventory shape:", inventory.shape)
```

```python
print("Sales columns:")
print(sales.columns.tolist())

print("\nProducts columns:")
print(products.columns.tolist())

print("\nInventory columns:")
print(inventory.columns.tolist())
```

```python
print("Sales sample:")
display(sales.head())

print("Products sample:")
display(products.head())

print("Inventory sample:")
display(inventory.head())
```

```python
datasets = {
    "Sales": sales,
    "Products": products,
    "Inventory": inventory
}

for name, df in datasets.items():
    print("\n" + "=" * 60)
    print(name.upper())
    print("=" * 60)

    print("Shape:", df.shape)
    print("Duplicate rows:", df.duplicated().sum())

    print("\nData types:")
    print(df.dtypes)

    missing = df.isna().sum()
    missing = missing[missing > 0]

    print("\nMissing values:")
    if missing.empty:
        print("No missing values")
    else:
        display(
            pd.DataFrame({
                "Missing Count": missing,
                "Missing Percentage": (missing / len(df) * 100).round(2)
            })
        )
```

```python
sales_dates = pd.to_datetime(sales["InvoiceDate"], errors="coerce")
invoice_text = sales["Invoice"].astype(str)

print("Sales date range:", sales_dates.min(), "to", sales_dates.max())
print("Invalid sales dates:", sales_dates.isna().sum())
print("Unique invoices:", sales["Invoice"].nunique())
print("Unique SKUs:", sales["StockCode"].nunique())
print("Unique customers:", sales["CustomerID"].nunique(dropna=True))

print("\nSales quality checks:")
print("Missing CustomerID:", sales["CustomerID"].isna().sum())
print("Missing StockCode:", sales["StockCode"].isna().sum())
print("Quantity <= 0:", (sales["Quantity"] <= 0).sum())
print("UnitPrice <= 0:", (sales["UnitPrice"] <= 0).sum())
print(
    "Invoice numbers beginning with C:",
    invoice_text.str.startswith("C", na=False).sum()
)
print("Exact duplicate sales rows:", sales.duplicated().sum())
```

```python
sales_skus = set(sales["StockCode"].dropna().astype(str))
product_skus = set(products["StockCode"].dropna().astype(str))

missing_product_skus = sales_skus - product_skus

print("Product records:", len(products))
print("Unique product SKUs:", products["StockCode"].nunique())
print("Duplicate StockCode values:", products["StockCode"].duplicated().sum())
print("Missing descriptions:", products["Description"].isna().sum())
print("Missing categories:", products["Category"].isna().sum())
print("UnitCost <= 0:", (products["UnitCost"] <= 0).sum())
print("UnitPrice <= 0:", (products["UnitPrice"] <= 0).sum())
print("LeadTimeDays <= 0:", (products["LeadTimeDays"] <= 0).sum())
print("Sales SKUs missing from product master:", len(missing_product_skus))

if missing_product_skus:
    print("Missing SKU examples:", list(missing_product_skus)[:10])
```

## Day 2 Data-Profiling Findings

### Sales Transactions

- The sales dataset contains 357,645 rows and 11 columns.
- It covers the period from 10 January 2023 to 30 December 2024.
- It contains 91,127 unique invoices, 150 SKUs and 7,603 identified customers.
- CustomerID is missing in 17,882 records, representing 5% of the dataset.
- There are 10,750 records with non-positive quantities.
- The same 10,750 records belong to invoices beginning with C, indicating cancellations.
- No invalid or non-positive unit prices were found.
- No exact duplicate sales rows were found.

### Product Master

- The product master contains 150 unique products.
- No duplicate product IDs were found.
- No missing descriptions or categories were found.
- No invalid unit costs, selling prices or lead times were found.
- Every sales SKU exists in the product master.

### Inventory

- The inventory dataset contains 109,650 rows and covers 1 January 2023 to 31 December 2024.
- It contains 150 unique SKUs.
- No invalid dates or duplicate Date-SKU records were found.
- Every inventory SKU exists in the product master.
- No negative numeric values or missing values were found.

### Proposed Cleaning Decisions

- Exclude cancellation transactions from fulfilled-demand calculations.
- Retain rows with missing CustomerID because customer identity is not required for SKU-level demand forecasting.
- Convert date columns to datetime format through code.
- Standardise StockCode values as strings.
- Preserve the original raw files without manual editing.

```python
inventory_dates = pd.to_datetime(inventory["Date"], errors="coerce")

product_skus = set(products["StockCode"].dropna().astype(str))
inventory_skus = set(inventory["StockCode"].dropna().astype(str))

missing_inventory_skus = inventory_skus - product_skus

print("Inventory date range:", inventory_dates.min(), "to", inventory_dates.max())
print("Invalid inventory dates:", inventory_dates.isna().sum())
print("Unique inventory SKUs:", inventory["StockCode"].nunique())

print(
    "Duplicate Date-SKU records:",
    inventory.duplicated(subset=["Date", "StockCode"]).sum()
)

print(
    "Inventory SKUs missing from product master:",
    len(missing_inventory_skus)
)

numeric_columns = inventory.select_dtypes(include="number").columns
negative_counts = (inventory[numeric_columns] < 0).sum()
negative_counts = negative_counts[negative_counts > 0]

print("\nNegative-value checks:")
if negative_counts.empty:
    print("No negative numeric values")
else:
    display(negative_counts.to_frame("Negative Count"))
```
